```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = "API_SP.POP.TOTL_DS2_en_csv_v2_38144.csv"
data = pd.read_csv(file_path, skiprows=4)

sample = data[["Country Name", "2020"]].head(10).copy()
sample["2020"] = pd.to_numeric(sample["2020"], errors="coerce")

plt.figure(figsize=(12, 6))
plt.bar(sample["Country Name"], sample["2020"])

plt.title("Population Distribution in 2020")
plt.xlabel("Country")
plt.ylabel("Population")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
